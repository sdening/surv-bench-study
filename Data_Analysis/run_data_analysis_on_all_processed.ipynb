```python
datasets = [
    "vlbw", "LeukSurv", "vdv", "Framingham", "support2", "gse3143",
    "prostateSurvival", "cost", "dataDIVAT1", "acath", "DLBCL",
    "pharmacoSmoking", "AML_Bull", "Dialysis", "cgd",
    "UnempDur", "chop", "Unemployment", "micro.censure", "cancer",
    "burn", "mimic_iv", "DBCD", "hdfail", "gse1992", "smarto", "Z243",
    "nwtco", "zinc", "dataDIVAT3", "gse4335", "pbc", "MCLcleaned", "e1684",
    "Aids2", "hepatoCellular", "kkbox_100k", "ovarian", "rhc", "flchain",
    "scania", "breast", "nki70", "NSBCD", "metabric_genes", "borrow_repay",
    "ADNI_dem_biom_hipp", "divorce", "veteran",
]

TIME_COL="time"
EVENT_COL="event"
```

```python
from datetime import datetime

preprocessed_data_dir = "../Data_Preprocessing/data/processed/"
summary_rows = []

for i, ds in enumerate(datasets, 1):
    processed_file = os.path.join(preprocessed_data_dir, ds, f"{ds}_processed.csv")
    print(f"-------- {ds} - {i} --------")
    if not os.path.exists(processed_file):
        print(f"  Preprocessed file not found: {processed_file}")
        continue
    print(f"Loading dataset: {processed_file}")
    df = pd.read_csv(processed_file)
    row = complete_analysis(df, ds_name=ds, TIME_COL="time", EVENT_COL="event")
    summary_rows.append(row)

summary_df = pd.DataFrame(summary_rows)
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
out_path = f"dataset_summary_stats_{timestamp}.csv"
summary_df.to_csv(out_path, index=False)
print(f"Saved summary stats for {len(summary_df)} datasets in {out_path}")
summary_df
```
